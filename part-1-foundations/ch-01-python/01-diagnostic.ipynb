```python
def myAvg(nums):
    if not nums: return None
    return sum(nums) / len(nums)

print(myAvg([10, 10, 10]))
```

```python
my_list = [{"nome": "ana", "nota": 7}, {"nome": "bruno", "nota": 4}, {"nome": "pedro", "nota": 2}, {"nome": "jose", "nota": 9}]
approved = [x["nome"] for x in my_list if x["nota"] >= 6]

print(approved)
```

```python
text = "o rato rato roeu a roupa do rei de roma roma roma roupa"
words = text.split()

word_counts = {}
for word in words:
    word_counts[word] = word_counts.get(word, 0) + 1

print(word_counts)
```

```python
class Counter:

    def __init__(self, value):
        self.value = value

    def increment(self, increment):
        self.value += increment

    def getValue(self):
        return self.value

counter = Counter(10)
counter2 = Counter(43654)

print(counter.getValue())

counter.increment(25)

print(counter.getValue())

print(counter2.getValue())
```
