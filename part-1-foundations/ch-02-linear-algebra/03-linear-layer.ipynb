```python
import numpy as np

rng = np.random.default_rng(42)

batch_size = 64
n_features = 10
n_outputs = 5

inputs = rng.normal(size=(batch_size, n_features))

weights = rng.normal(size=(n_features, n_outputs))
bias = np.zeros(n_outputs)

outputs = inputs @ weights + bias

print(inputs.shape, weights.shape, bias.shape, outputs.shape)
assert outputs.shape == (batch_size, n_outputs)

outputs_loop = np.zeros((batch_size, n_outputs))

for i in range(batch_size):
    for j in range(n_outputs):
        total = 0.0
        for f in range(n_features):
            total += inputs[i, f] * weights[f, j]
            
        outputs_loop[i, j] = total + bias[j]

assert np.allclose(outputs, outputs_loop)
print("equal")
```
